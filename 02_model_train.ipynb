```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("cleaned.csv")
```

```python
df.head()
```

```python
df.tail()
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv = CountVectorizer(max_features=5000, stop_words="english")
```

```python
vectors = cv.fit_transform(df.tags).toarray()
```

```python
vectors.shape
```

```python
cv.get_feature_names_out()
```

```python
cv.vocabulary_
```

```python
vectors[0]
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(vectors)
```

```python
similarity.shape
```

```python
similarity[0]
```

```python
def recommender(movies_list):
    new_list = [movie.lower() for movie in movies_list]
    recommendation = []
    for mov in new_list:
        index = df[df["title"].str.lower() == mov].index[0]
        distance = sorted(list(enumerate(similarity[index])),reverse=True, key=lambda x:x[1])
        for i in distance[0:4]:
            recommendation.append(df.iloc[i[0]].title)
    return recommendation
```

```python
recommender(["Avatar", "Titanic"])
```

```python
#export the file for the real world applications

import pickle
pickle.dump(similarity,open("similarity.pkl", "wb"))
```
